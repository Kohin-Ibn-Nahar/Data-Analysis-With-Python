```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_excel("/kaggle/input/datasets/kohinrahman/kohinrahman/Air_BNB.xlsx")
```

```python
data.drop('id',axis=1,inplace=True)
```

```python
num = []
cat = []
bl = []
for i in data.columns:
    if data[i].dtype=='object':
        cat.append(i)
    elif data[i].dtype=='bool':
        bl.append(i)
    else:
        num.append(i)
```

```python
num
```

```python
dependent_var = 'log_price'
```

```python
num.remove(dependent_var)
```

```python
num
```

```python
t = data[num].corr()
```

```python
t
```

```python
sns.heatmap(data[num].corr(),annot = True,fmt='.2g')
```

```python
threshold = 0.70
selected_feature = num.copy()

for i in range(t.shape[0]):
    for j in range(i + 1, t.shape[1]):
        if abs(t.values[i, j]) >= threshold:
            feature = t.columns[j]
            if feature in selected_feature:
                selected_feature.remove(feature)

print(selected_feature)
```

```python
sns.heatmap(data[selected_feature].corr(),annot = True,fmt = '.2g')
```
